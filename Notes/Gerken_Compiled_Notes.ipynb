# Compiled Gerken Notes

This notebook consolidates the project notes related to data availability, Massanutten snowfall, Daymet, and the later gridded-data follow-up.


## Long-Record Data Direction

- Use longer datasets from Daymet.
- Take the roughly 10 years of local data already in hand and align it with the longer Daymet dataset.
- Compare how well the local observations and the longer gridded dataset agree.
- If they agree well, that is useful. If they do not agree well, that is also useful because it tells you something about the limits of the gridded product.


## Valley-to-Mountain Comparison

- Look at spatial differences from Harrisonburg up the mountain.
- Examine how precipitation and temperature change from the valley to Massanutten.
- Look at the correlation between precipitation and sub-zero temperatures.
- Think about the seasonality of precipitation and when it matters most for snow conditions.


## Snow-Focused Interpretation

- If the gridded dataset does not give direct snowfall cleanly, focus on when the precipitation falls and whether temperatures are cold enough for snow.
- Count how many days per year are below freezing.
- Degree days can be useful as a summary of seasonal cold conditions.
- Compare observed local data to the spatial dataset to see how much more or less the mountain is receiving than nearby lower-elevation locations.


## Gridded Mapping Follow-Up

- Add a small gridded map product for Massanutten.
- A 30 km box is enough for a first test.
- Build a small NetCDF subset if possible rather than downloading broad products.
- Daymet is the first choice if a gridded subset can be obtained cleanly.
- If Daymet is awkward for the gridded subset, ERA5 is a fallback, but only download exactly what is needed because the files get large quickly.


## Scope Guidance For Gridded Downloads

- Keep the first map very small and specific.
- December is a good first month because it is snow-relevant and keeps the file size manageable.
- Use only the variables that directly support the question.
- For Daymet, the priority variables are `prcp`, `tmin`, `tmax`, and `swe` if available.
- Start with one month and one small region before scaling up to multiple years or a full winter season.


## Practical Project Structure

A practical structure implied by the notes is:
1. Use the long-term NOAA proxy station for long snowfall and temperature trends.
2. Use the recent Massanutten station for local observed snowfall and precipitation.
3. Use Daymet point data for long valley-versus-mountain precipitation and temperature comparisons.
4. Add a small gridded December subset to visualize the mountain-to-valley gradient spatially.
